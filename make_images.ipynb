```python
import pandas as pd
import chess.pgn
import io
import cairosvg

openings = pd.read_csv("vertices.csv")
```

```python
for i, row in openings.iterrows():
    opening = row["name"]
    
    game = chess.pgn.read_game(io.StringIO(row['moves'].replace('.', '. ')))
    board = game.board()
    for move in game.mainline_moves():
        board.push(move)
    
    board_svg = chess.svg.board(board, colors = {"square dark": "#759655", "square light": "#eeeed1"}, size=600)
    cairosvg.svg2png(board_svg, write_to=f'./images/{opening}_white.png')

    board_svg = chess.svg.board(board, colors = {"square dark": "#759655", "square light": "#eeeed1"}, size=600, orientation=chess.BLACK)
    cairosvg.svg2png(board_svg, write_to=f'./images/{opening}_black.png')
```
